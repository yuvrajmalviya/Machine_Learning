```python
df = pd.read_csv('cars.csv', usecols=['km_driven', 'selling_price'])
```

```python
df
```

```python
import seaborn as sns
```

```python
sns.histplot(df['km_driven'], kde=True)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    df.iloc[:,:],
    df['selling_price'],test_size=0.2
)
```

```python
x_train
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import FunctionTransformer
trf = ColumnTransformer(transformers = [('logs', FunctionTransformer(func=np.log1p), [0])], remainder='passthrough')
```

```python
x_train_transformed = trf.fit_transform(x_train)
x_test_transformed = trf.transform(x_test)
```

```python
x_train_transformed
```

```python
sns.distplot(x_train_transformed)
```
