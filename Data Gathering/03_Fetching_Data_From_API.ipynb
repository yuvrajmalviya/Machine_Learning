```python
import requests
import pandas as pd

url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": 52.52,
    "longitude": 13.418,
    "start_date": "2021-01-01",
    "end_date": "2021-01-03",
    "hourly": "temperature_2m"
}

response = requests.get(url, params=params)

data = response.json()

df = pd.DataFrame(data)
df
```
