### **1. Scatterplot (Numerical - Numerical)**

```python
# bivariate
sns.scatterplot(data=tips, x='total_bill', y='tip')
```

```python
# multivariate
sns.scatterplot(data=tips, x='total_bill',y='tip', hue='sex', style='smoker', size='size')
```

### **2. Barplot (Numerical - Categorical)**

```python
titanic.head()
```

```python
sns.barplot(data=titanic, x='Pclass', y='Fare')
```

```python
sns.barplot(data=titanic, x='Pclass', y='Fare', hue='Sex')
```

### **3. Boxplot (Numerical - Categorical)**

```python
sns.boxplot(data=titanic, x='Sex',y='Age', hue='Survived')
```

### **4. Displot (Numerical - Categorical)**

```python
sns.distplot(titanic[titanic['Survived']==0]['Age'], hist=False)
sns.distplot(titanic[titanic['Survived']==1]['Age'], hist=False)
```

```python
sns.displot(data=titanic, x='Age', hue='Survived', kind='kde')
```

```python
sns.histplot(titanic[titanic['Survived']==0]['Age'], kde=True)
sns.histplot(titanic[titanic['Survived']==1]['Age'],kde=True)
```

```python
sns.histplot(
    data=titanic,
    x='Age',
    hue='Survived',
    kde=True
)
plt.show()
```

### **5. HeatMap (Categorical - Categorical)**

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```

```python
(titanic.groupby('Embarked')['Survived'].mean()*100).plot(kind='bar')
```

```python
(titanic.groupby('Pclass')['Survived'].mean()*100).plot(kind='bar')
```

### **6. ClusterMap (Categorical - Categorical)**

```python
sns.clustermap(
    pd.crosstab(titanic['SibSp'], titanic['Survived'])
)

plt.show()
```
