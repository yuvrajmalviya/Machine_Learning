```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('D:/Machine_Learning/Data Gathering/Titanic-Dataset.csv')
```

```python
df.head()
```

## **1. Categorical Data**

### a. Countplot

```python
sns.countplot(x=df['Survived'], palette=['red','green'])
```

```python
df['Survived'].value_counts().plot(kind='bar', color=['Red','Green'])
```

```python
# Pclass
sns.countplot(data=df, x='Pclass', hue='Pclass')
```

```python
# Embarked
sns.countplot(data=df, x='Embarked', hue='Embarked')
```

#### b. PieChart

```python
df['Pclass'].value_counts().plot(kind='pie', autopct='%.2f%%')
```

```python
df['Sex'].value_counts().plot(kind='pie', autopct='%.2f%%')
```


### **2. Numerical Data**


### a. Histogram

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Age'], bins=20)
```

### b. Distplot

```python
sns.displot(data = df, x="Age", kind='kde')
```

```python
sns.displot(data=df, x='Age', kind='ecdf')
```

### c. Boxplot

```python
sns.boxplot(data=df, x='Fare')
```
