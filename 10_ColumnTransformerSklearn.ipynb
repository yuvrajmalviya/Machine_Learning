```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
covid = pd.read_csv('covid_toy.csv')
```

```python
covid.sample(5)
```

```python
covid.shape
```

```python
covid.isnull().sum()
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    covid.drop(columns=['has_covid']),
    covid['has_covid'],
    test_size=0.2,
)
```

```python
x_train
```

### Without Using ColumnTransformer

```python
from sklearn.impute import SimpleImputer
si = SimpleImputer()
```

```python
x_train_fever = si.fit_transform(x_train[['fever']])
x_test_fever = si.fit_transform(x_test[['fever']])
```

```python
x_train_fever
```

```python
# OrdinalEndoder on cough column
oe = OrdinalEncoder(categories=[['Mild','Strong']], dtype=int)
```

```python
x_train_cough = oe.fit_transform(x_train[['cough']])
x_test_cough = oe.fit_transform(x_test[['cough']])
```

```python
x_train_cough.shape
```

```python
# OneHotEncoder on gender & city
ohe = OneHotEncoder(drop='first', sparse_output=False, dtype=int)
```

```python
x_train_ordinal = ohe.fit_transform(x_train[['gender', 'city']])
x_test_ordinal = ohe.fit_transform(x_test[['gender', 'city']])
```

```python
x_train_ordinal.shape
```

```python
x_test_ordinal.shape
```

```python
# Extract age from covid dataset
x_train_age = x_train.drop(columns = ['gender', 'fever', 'cough', 'city'])
x_test_age = x_test.drop(columns=['gender', 'fever', 'cough', 'city'])
```

```python
x_train_age.shape
```

```python
# concatenate
x_train_transform = np.concatenate((x_train_age, x_train_fever,x_train_ordinal, x_train_cough), axis=1)
x_test_transform = np.concatenate((x_test_age,x_test_fever,x_test_ordinal,x_test_cough), axis=1)
```

```python
x_train_transform.shape
```

### Using ColumnTransformer

```python
from sklearn.compose import ColumnTransformer

transformer = ColumnTransformer(transformers=[
    ('tnf1', SimpleImputer(), ['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Mild','Strong']], dtype=int), ['cough']),
    ('tnf3', OneHotEncoder(drop='first', sparse_output=False, dtype=int), ['gender', 'city'])
], remainder='passthrough')
```

```python
x_train_transformed = transformer.fit_transform(x_train)
```

```python
 x_test_transformed = transformer.fit_transform(x_test)
```
